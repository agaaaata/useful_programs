## Suma ciągu geometrycznego

Rozważmy ciąg geometryczny o pierwszym wyrazie $a_1$ oraz ilorazie $q$.

Suma pierwszych $n$ wyrazów wynosi:

$$
S_n = a_1 \frac{1-q^n}{1-q}
$$

dla $q \neq 1$.

Przykład:

Dla

$$
a_1 = 2,\quad q = 3,\quad n = 5
$$

otrzymujemy

$$
S_5 = 2 \cdot \frac{1-3^5}{1-3}
= 242.
$$

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import Video
from matplotlib import animation

g = 9.81
L = 1.0

t = np.linspace(0, 10, 300)

theta = 0.4 * np.cos(np.sqrt(g/L) * t)

x = L * np.sin(theta)
y = -L * np.cos(theta)

fig, ax = plt.subplots(figsize=(5,5))
ax.set_xlim(-1.2, 1.2)
ax.set_ylim(-1.2, 0.2)
ax.set_aspect('equal')

line, = ax.plot([], [], 'o-', lw=2)

def update(frame):
    line.set_data([0, x[frame]], [0, y[frame]])
    return line,

ani = FuncAnimation(
    fig,
    update,
    frames=len(t),
    interval=30,
    blit=True
)

writer = animation.FFMpegWriter(fps=30)

ani.save("video.mp4", writer=writer)

plt.close()

Video("video.mp4", embed=True)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Parametr
t = np.linspace(0, 8 * np.pi, 1000)

# Nowa krzywa parametryczna - helisa
x = np.cos(t)
y = np.sin(t)
z = t / (2 * np.pi)

# Rysowanie
fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')

ax.plot(x, y, z, color='blue', linewidth=2, label='Helisa')

ax.set_title('Krzywa parametryczna - Helisa')
ax.set_xlabel('x(t)')
ax.set_ylabel('y(t)')
ax.set_zlabel('z(t)')
ax.legend()

# Zapis do pliku
plt.savefig("parametric_curve.png", dpi=300, bbox_inches="tight")

# Wyświetlenie
plt.show()

print("Zapisano plik: parametric_curve.png")
```
