# Sumowanie ciągu geometrycznego

Rozważmy ciąg geometryczny o pierwszym wyrazie $a_1$ oraz ilorazie $q$.

Suma pierwszych $n$ wyrazów ciągu geometrycznego wyraża się wzorem:

$$
S_n = a_1 \frac{1-q^n}{1-q}
$$

dla warunku:

```text
q ≠ 1
```

Przykład:

$$
a_1 = 2,\quad q = 3,\quad n = 4
$$

Wtedy:

$$
S_4 = 2\cdot\frac{1-3^4}{1-3} = 80
$$

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import Video

L = 1.0
g = 9.81

fig, ax = plt.subplots(figsize=(5,5))

ax.set_xlim(-1.2, 1.2)
ax.set_ylim(-1.2, 0.2)
ax.set_aspect('equal')
ax.grid()

line, = ax.plot([], [], 'o-', lw=3)

def update(frame):
    theta = 0.6 * np.cos(np.sqrt(g/L) * frame / 20)

    x = L * np.sin(theta)
    y = -L * np.cos(theta)

    line.set_data([0, x], [0, y])

    return line,

ani = FuncAnimation(
    fig,
    update,
    frames=200,
    interval=50,
    blit=True
)

ani.save('/content/video.mp4', writer='ffmpeg', fps=30)

plt.close()

Video('/content/video.mp4', embed=True)
````

````python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 10*np.pi, 2000)

x = t * np.cos(t) / 10
y = t * np.sin(t) / 10
z = t / 5

fig = plt.figure(figsize=(8,8))
ax = fig.add_subplot(111, projection='3d')

ax.plot(x, y, z, color='red', label='Conical Spiral')

ax.set_title("Conical Spiral")
ax.set_xlabel("x(t)")
ax.set_ylabel("y(t)")
ax.set_zlabel("z(t)")
ax.legend()

plt.savefig("/content/parametric_curve.png", dpi=300)
plt.show()
````
